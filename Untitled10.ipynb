```python
import pandas as pd
import math
import matplotlib.pyplot as plt
```

```python
import pandas as pd
from google.colab import files
uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_excel(filename)
print(df.head())
print(df.columns.tolist())
```

```python
df = df[["avg_temp", "rainfall"]]
df = df.dropna()
print(df.head())
print("Number of records:", len(df))
```

```python
def correlation(x, y):
    n = len(x)
    mean_x = sum(x) / n
    mean_y = sum(y) / n
    numerator = sum(
        (x[i] - mean_x) * (y[i] - mean_y)
        for i in range(n)
    )
    sum_x = sum((value - mean_x) ** 2 for value in x)
    sum_y = sum((value - mean_y) ** 2 for value in y)

    denominator = math.sqrt(sum_x * sum_y)

    if denominator == 0:
        return float("nan")

    return numerator / denominator
```

```python

x = df["avg_temp"].tolist()
y = df["rainfall"].tolist()
r = correlation(x, y)

print("Variables: Average Temperature and Rainfall")
print("Number of records:", len(x))
print("Correlation coefficient:", round(r, 4))
```

```python
if math.isnan(r):
    print("Correlation cannot be calculated.")
elif r > 0:
    print("Positive correlation")
elif r < 0:
    print("Negative correlation")
else:
    print("No linear correlation")
```

```python
plot_data = df.sample(
    n=min(3000, len(df)),
    random_state=42
)

plt.scatter(
    plot_data["avg_temp"],
    plot_data["rainfall"]
)

plt.xlabel("Average Temperature")
plt.ylabel("Rainfall")
plt.title("Correlation Between Average Temperature and Rainfall")
plt.grid(True)
plt.show()
```

```python

x_positive = [1, 2, 3, 4, 5]
y_positive = [2, 4, 6, 8, 10]

x_negative = [1, 2, 3, 4, 5]
y_negative = [10, 8, 6, 4, 2]

x_zero = [1, 2, 3, 4, 5]
y_zero = [1, -1, 1, -1, 0]
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.scatter(x_positive, y_positive)
plt.title("Positive Correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.subplot(1, 3, 2)
plt.scatter(x_negative, y_negative)
plt.title("Negative Correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.subplot(1, 3, 3)
plt.scatter(x_zero, y_zero)
plt.title("No Linear Correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.tight_layout()
plt.show()
```
